```python
import random

def fitness_function(position):
    """
    The mathematical function we want to minimize (Sphere Function).
    Target minimum is 0 at.
    """
    return sum(x**2 for x in position)

def run_pso(num_particles=30, dimensions=2, max_iterations=50):
    # Hyperparameters
    w = 0.5   # Inertia weight (how much to keep current speed)
    c1 = 1.5  # Cognitive coefficient (pull towards personal best)
    c2 = 1.5  # Social coefficient (pull towards global best)

    # 1. INITIALIZATION
    positions = []
    velocities = []
    pbest_positions = []
    pbest_fitnesses = []

    gbest_position = None
    gbest_fitness = float('inf')

    for i in range(num_particles):
        # Initialize randomly between -10 and 10
        pos = [random.uniform(-10, 10) for _ in range(dimensions)]
        vel = [random.uniform(-1, 1) for _ in range(dimensions)]

        positions.append(pos)
        velocities.append(vel)
        pbest_positions.append(list(pos))

        # Evaluate initial fitness
        fitness = fitness_function(pos)
        pbest_fitnesses.append(fitness)

        # Check if this is the global best so far
        if fitness < gbest_fitness:
            gbest_fitness = fitness
            gbest_position = list(pos)

    # 2. OPTIMIZATION LOOP
    for iteration in range(max_iterations):
        for i in range(num_particles):
            for d in range(dimensions):
                r1 = random.random()
                r2 = random.random()

                # --- The Core PSO Math Equations ---
                # Update velocity (speed)
                velocities[i][d] = (w * velocities[i][d] +
                                    c1 * r1 * (pbest_positions[i][d] - positions[i][d]) +
                                    c2 * r2 * (gbest_position[d] - positions[i][d]))

                # Update position (location)
                positions[i][d] += velocities[i][d]

            # Evaluate new position
            fitness = fitness_function(positions[i])

            # Update Personal Best
            if fitness < pbest_fitnesses[i]:
                pbest_fitnesses[i] = fitness
                pbest_positions[i] = list(positions[i])

            # Update Global Best
            if fitness < gbest_fitness:
                gbest_fitness = fitness
                gbest_position = list(positions[i])

        # Optional: Print progress every 10 iterations
        if (iteration + 1) % 10 == 0:
            print(f"Iteration {iteration + 1}/{max_iterations} - Best Fitness: {gbest_fitness:.6f}")

    return gbest_position, gbest_fitness

# --- Run the Algorithm ---
if __name__ == "__main__":
    best_coordinates, best_score = run_pso(num_particles=30, dimensions=2, max_iterations=50)
    print("\n--- Optimization Complete ---")
    print(f"Best Found Coordinates: {[round(c, 4) for c in best_coordinates]}")
    print(f"Best Fitness Value: {best_score:.6f}")
```
